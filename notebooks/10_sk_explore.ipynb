```python
import pandas as pd

path = '/content/drive/MyDrive/raw/sk/flow_age_pop_202507.csv'

df = pd.read_csv(path, sep='|')

# BLOCK_CD를 문자열로 변환
df['BLOCK_CD'] = df['BLOCK_CD'].astype(str)

# 강남구 / 춘천시만 추출
df_gangnam_chuncheon = df[
    df['BLOCK_CD'].str[:5].isin(['11230', '32010'])
].copy()

print("전체 데이터:", len(df))
print("강남구 + 춘천시:", len(df_gangnam_chuncheon))

# 지역 구분
df_gangnam_chuncheon['REGION'] = df_gangnam_chuncheon['BLOCK_CD'].str[:5].map({
    '11230': '강남구',
    '32010': '춘천시'
})

print(df_gangnam_chuncheon['REGION'].value_counts())
```
