```python
import pandas as pd
import glob
import os

# 파일이 있는 실제 폴더
base_path = "/content/drive/MyDrive/bigcontest/data/sk_processed"

# 강남 / 춘천 파일 자동 검색
gangnam_files = glob.glob(os.path.join(base_path, "flow_*_gangnam_202507_202512.csv"))
chuncheon_files = glob.glob(os.path.join(base_path, "flow_*_chuncheon_202507_202512.csv"))

print("강남 파일:")
for f in gangnam_files:
    print(os.path.basename(f))

print("\n춘천 파일:")
for f in chuncheon_files:
    print(os.path.basename(f))


# 좌표 추출 함수
def extract_coordinates(files):

    dfs = []

    for file in files:

        df = pd.read_csv(file, encoding="utf-8-sig")

        # 필요한 컬럼만 추출
        coord = df[
            ["STD_YM", "BLOCK_CD", "X_COORD", "Y_COORD"]
        ].copy()

        dfs.append(coord)

    # 파일 합치기
    result = pd.concat(dfs, ignore_index=True)

    # 팀원이 말한 키 기준으로 중복 제거
    result = result.drop_duplicates(
        subset=[
            "STD_YM",
            "BLOCK_CD",
            "X_COORD",
            "Y_COORD"
        ]
    )

    return result


# 강남 / 춘천 좌표 생성
gangnam_coord = extract_coordinates(gangnam_files)
chuncheon_coord = extract_coordinates(chuncheon_files)


# 저장
gangnam_path = os.path.join(
    base_path,
    "SK_coordinate_gangnam.csv"
)

chuncheon_path = os.path.join(
    base_path,
    "SK_coordinate_chuncheon.csv"
)

gangnam_coord.to_csv(
    gangnam_path,
    index=False,
    encoding="utf-8-sig"
)

chuncheon_coord.to_csv(
    chuncheon_path,
    index=False,
    encoding="utf-8-sig"
)


# 결과 확인
print("\n====================")
print("완료!")
print("====================")

print("강남 좌표:", len(gangnam_coord), "행")
print("춘천 좌표:", len(chuncheon_coord), "행")

print("\n강남 파일:")
print(gangnam_path)

print("\n춘천 파일:")
print(chuncheon_path)

print("\n강남 미리보기")
display(gangnam_coord.head())

print("\n춘천 미리보기")
display(chuncheon_coord.head())
```
