```python
import pandas as pd
import os

# SK 유동인구 데이터 폴더
folder = '/content/drive/MyDrive/raw/sk'

# 2025년 7월 ~ 12월
months = ['202507', '202508', '202509', '202510', '202511', '202512']

age_dfs = []

for month in months:
    path = os.path.join(folder, f'flow_age_pop_{month}.csv')

    print(f'{month} 파일 읽는 중...')

    df = pd.read_csv(path, sep='|')

    # BLOCK_CD를 문자열로 변환
    df['BLOCK_CD'] = df['BLOCK_CD'].astype(str)

    # 강남구 + 춘천시만 추출
    df = df[
        df['BLOCK_CD'].str[:5].isin(['11230', '32010'])
    ].copy()

    # 지역명 추가
    df['REGION'] = df['BLOCK_CD'].str[:5].map({
        '11230': '강남구',
        '32010': '춘천시'
    })

    age_dfs.append(df)

# 6개월 데이터 합치기
age_gangnam_chuncheon = pd.concat(age_dfs, ignore_index=True)

print('\n===== 완료 =====')
print('전체 행 수:', len(age_gangnam_chuncheon))
print('\n지역별 행 수:')
print(age_gangnam_chuncheon['REGION'].value_counts())

print('\n월별 행 수:')
print(age_gangnam_chuncheon['STD_YM'].value_counts().sort_index())
```

```python
output_path = '/content/drive/MyDrive/raw/flow_age_gangnam_chuncheon_202507_202512.csv'

age_gangnam_chuncheon.to_csv(
    output_path,
    index=False,
    encoding='utf-8-sig'
)

print("저장 완료!")
print(output_path)
```

```python
import pandas as pd
import os

folder = '/content/drive/MyDrive/raw/sk'
months = ['202507', '202508', '202509', '202510', '202511', '202512']

wkdy_dfs = []

for month in months:
    path = os.path.join(folder, f'flow_wkdy_pop_{month}.csv')

    print(f'{month} 파일 읽는 중...')

    df = pd.read_csv(path, sep='|')
    df['BLOCK_CD'] = df['BLOCK_CD'].astype(str)

    df = df[
        df['BLOCK_CD'].str[:5].isin(['11230', '32010'])
    ].copy()

    df['REGION'] = df['BLOCK_CD'].str[:5].map({
        '11230': '강남구',
        '32010': '춘천시'
    })

    wkdy_dfs.append(df)

wkdy_gangnam_chuncheon = pd.concat(
    wkdy_dfs,
    ignore_index=True
)

print('\n===== 완료 =====')
print('전체 행 수:', len(wkdy_gangnam_chuncheon))

print('\n지역별 행 수:')
print(wkdy_gangnam_chuncheon['REGION'].value_counts())

print('\n월별 행 수:')
print(
    wkdy_gangnam_chuncheon['STD_YM']
    .value_counts()
    .sort_index()
)
```

```python
import os

# 저장할 폴더
base_folder = '/content/drive/MyDrive/raw/sk_processed'

gangnam_folder = os.path.join(base_folder, 'gangnam')
chuncheon_folder = os.path.join(base_folder, 'chuncheon')

os.makedirs(gangnam_folder, exist_ok=True)
os.makedirs(chuncheon_folder, exist_ok=True)

# =========================
# 1. 나이·성별별 유동인구
# =========================

age_gangnam = age_gangnam_chuncheon[
    age_gangnam_chuncheon['REGION'] == '강남구'
].copy()

age_chuncheon = age_gangnam_chuncheon[
    age_gangnam_chuncheon['REGION'] == '춘천시'
].copy()


# =========================
# 2. 시간대별 유동인구
# =========================

time_gangnam = time_gangnam_chuncheon[
    time_gangnam_chuncheon['REGION'] == '강남구'
].copy()

time_chuncheon = time_gangnam_chuncheon[
    time_gangnam_chuncheon['REGION'] == '춘천시'
].copy()


# =========================
# 3. 요일별 유동인구
# =========================

wkdy_gangnam = wkdy_gangnam_chuncheon[
    wkdy_gangnam_chuncheon['REGION'] == '강남구'
].copy()

wkdy_chuncheon = wkdy_gangnam_chuncheon[
    wkdy_gangnam_chuncheon['REGION'] == '춘천시'
].copy()


# =========================
# 4. 각각 CSV로 저장
# =========================

age_gangnam.to_csv(
    f'{gangnam_folder}/flow_age_gangnam_202507_202512.csv',
    index=False,
    encoding='utf-8-sig'
)

time_gangnam.to_csv(
    f'{gangnam_folder}/flow_time_gangnam_202507_202512.csv',
    index=False,
    encoding='utf-8-sig'
)

wkdy_gangnam.to_csv(
    f'{gangnam_folder}/flow_wkdy_gangnam_202507_202512.csv',
    index=False,
    encoding='utf-8-sig'
)

age_chuncheon.to_csv(
    f'{chuncheon_folder}/flow_age_chuncheon_202507_202512.csv',
    index=False,
    encoding='utf-8-sig'
)

time_chuncheon.to_csv(
    f'{chuncheon_folder}/flow_time_chuncheon_202507_202512.csv',
    index=False,
    encoding='utf-8-sig'
)

wkdy_chuncheon.to_csv(
    f'{chuncheon_folder}/flow_wkdy_chuncheon_202507_202512.csv',
    index=False,
    encoding='utf-8-sig'
)

print("✅ 6개 데이터 저장 완료!")
print(f"강남구 - 나이: {len(age_gangnam):,}행")
print(f"강남구 - 시간: {len(time_gangnam):,}행")
print(f"강남구 - 요일: {len(wkdy_gangnam):,}행")
print(f"춘천시 - 나이: {len(age_chuncheon):,}행")
print(f"춘천시 - 시간: {len(time_chuncheon):,}행")
print(f"춘천시 - 요일: {len(wkdy_chuncheon):,}행")
```

```python
import pandas as pd
import os

folder = '/content/drive/MyDrive/bigcontest/data/sk_processed'

files = [
    'flow_age_chuncheon_202507_202512.csv',
    'flow_age_gangnam_202507_202512.csv',
    'flow_time_chuncheon_202507_202512.csv',
    'flow_time_gangnam_202507_202512.csv',
    'flow_wkdy_chuncheon_202507_202512.csv',
    'flow_wkdy_gangnam_202507_202512.csv'
]

for file in files:

    path = os.path.join(folder, file)

    # 기존 파일 읽기1
    df = pd.read_csv(path, encoding='utf-8-sig')

    before = len(df)

    # 완전히 동일한 행 제거
    df = df.drop_duplicates()

    after = len(df)

    # ★ 기존 파일에 그대로 덮어쓰기
    df.to_csv(path, index=False, encoding='utf-8-sig')

    print(f'📁 {file}')
    print(f'중복 제거 전 : {before:,}행')
    print(f'중복 제거 후 : {after:,}행')
    print(f'제거된 행    : {before-after:,}행')
    print(f'남은 중복    : {df.duplicated().sum()}')
    print('-' * 50)
```
